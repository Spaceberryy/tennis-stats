```python
import pandas as pd
```

```python
df =  pd.read_csv("../data/UTS/federer.csv")
```

```python
df
```

```python
match_results = list(df['winner_name'] == 'Roger Federer')
```

```python
win_count = 0
for win in match_results:
    if win == True:
        win_count += 1
```

```python
win_count
```

```python
win_percentage = (win_count / len(match_results)) * 100
print(win_percentage)
```
