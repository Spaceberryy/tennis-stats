```python
def get_ace_count(data, player_name):
    aces_in_match_won = data.loc[data['winner_name'] == player_name, 'w_ace'].sum()
    aces_in_match_lost = data.loc[data['loser_name'] == player_name, 'l_ace'].sum()

    total_aces = aces_in_match_won + aces_in_match_lost

    return total_aces


print(get_ace_count(data, 'Carlos Alcaraz'))
print(get_ace_count(data, 'Jannik Sinner'))
print(get_ace_count(data, 'Novak Djokovic'))
```

```python
def get_first_serve_percentage(data, player_name):
    total_service_points_match_won = data.loc[data['winner_name'] == player_name, 'w_svpt'].sum()
    total_first_serves_in_match_won = data.loc[data['winner_name'] == player_name, 'w_1stIn'].sum()

    total_service_points_match_lost = data.loc[data['loser_name'] == player_name, 'l_svpt'].sum()
    total_first_serves_in_match_lost = data.loc[data['loser_name'] == player_name, 'l_1stIn'].sum()

    first_serve_percentage = ((total_first_serves_in_match_lost + total_first_serves_in_match_won) / (total_service_points_match_won + total_service_points_match_lost)) * 100

    return first_serve_percentage

print(get_first_serve_percentage(data, 'Jannik Sinner'))
```

```python
# potential error: ff data has any matches with NaN in the bp columns (retirements, incomplete stats), .
#                  sum() will silently skip them, but your denominator could still be off if faced is present but saved is 
#                  missing (or vice versa) — worth a dropna or a sanity check before trusting this at scale.
def get_break_points_saved_rate(data, player_name):
    total_break_points_faced_in_matches_won = data.loc[data['winner_name'] == player_name, 'w_bpFaced'].sum()
    total_break_points_saved_in_matches_won = data.loc[data['winner_name'] == player_name, 'w_bpSaved'].sum()

    total_break_points_faced_in_matches_lost = data.loc[data['loser_name'] == player_name, 'l_bpFaced'].sum()
    total_break_points_saved_in_matches_lost = data.loc[data['loser_name'] == player_name, 'l_bpSaved'].sum()

    break_points_saved = ((total_break_points_saved_in_matches_won + total_break_points_saved_in_matches_lost) / (total_break_points_faced_in_matches_won + total_break_points_faced_in_matches_lost)) * 100

    return break_points_saved

print(get_break_points_saved_rate(data, 'Jannik Sinner'))
```

```python
def get_surface_win_rate(data, player_name, surface):
    total_surface_matches_won = len(data[(data['winner_name'] == player_name) & (data['surface'] == surface)])
    total_surface_matches_lost = len(data[(data['loser_name'] == player_name) & (data['surface'] == surface)])

    # TODO: Will need to consider division by zero error here. Will Fix later if need be.
    surface_win_rate = (total_surface_matches_won / (total_surface_matches_won + total_surface_matches_lost)) * 100

    return surface_win_rate

print(get_surface_win_rate(data, 'Jannik Sinner', 'Grass'))
```

```python
def head_to_head_record(data, player_one, player_two):
    no_matches_won_by_player_one = len(data[(data['winner_name'] == player_one) & (data['loser_name'] == player_two)])
    no_matches_won_by_player_two = len(data[(data['winner_name'] == player_two) & (data['loser_name'] == player_one)])

    return no_matches_won_by_player_one, no_matches_won_by_player_two

player_one_wins, player_two_wins = head_to_head_record(data, 'Jannik Sinner', 'Carlos Alcaraz')
print(f'{player_one_wins} - {player_two_wins}')
```

```python
def get_break_points_conversion_rate(data, player_name):
    # formula for calculating breaking points won :-
    #     if the player won the match: loser_break_points_faced - loser_break_points_saved
    #     if the player lost the match: winner_break_points_faced - winner_break_points_saved
    break_points_faced_by_opponent_when_winning = data.loc[data['winner_name'] == player_name, 'l_bpFaced'].sum()
    break_points_saved_by_opponent_when_winning = data.loc[data['winner_name'] == player_name, 'l_bpSaved'].sum()

    break_points_faced_by_opponent_when_losing = data.loc[data['loser_name'] == player_name, 'w_bpFaced'].sum()
    break_points_saved_by_opponent_when_losing = data.loc[data['loser_name'] == player_name, 'w_bpSaved'].sum()

    bp_won = (break_points_faced_by_opponent_when_losing + break_points_faced_by_opponent_when_winning) - (break_points_saved_by_opponent_when_losing + break_points_saved_by_opponent_when_winning)
    total_bp = break_points_faced_by_opponent_when_losing + break_points_faced_by_opponent_when_winning

    bp_conversion_rate = (bp_won / total_bp) * 100

    return bp_conversion_rate


print(get_break_points_conversion_rate(data, 'Jannik Sinner'))
```

```python
def get_first_serve_win_percentage(data, player_name):
    first_serve_points_match_won = data.loc[data['winner_name'] == player_name, 'w_1stIn'].sum()
    first_serves_won_in_match_won = data.loc[data['winner_name'] == player_name, 'w_1stWon'].sum()

    first_serve_points_match_lost = data.loc[data['loser_name'] == player_name, 'l_1stIn'].sum()
    first_serves_won_in_match_lost = data.loc[data['loser_name'] == player_name, 'l_1stWon'].sum()

    first_serve_win_percentage = ((first_serves_won_in_match_lost + first_serves_won_in_match_won) / (first_serve_points_match_won + first_serve_points_match_lost)) * 100

    return first_serve_win_percentage

print(get_first_serve_win_percentage(data, 'Jannik Sinner'))
```

```python
def get_second_serve_win_percentage(data, player_name):
    first_serve_points_match_won = data.loc[data['winner_name'] == player_name, 'w_1stIn'].sum()
    second_serves_won_in_match_won = data.loc[data['winner_name'] == player_name, 'w_2ndWon'].sum()
    total_service_points_match_won = data.loc[data['winner_name'] == player_name, 'w_svpt'].sum()

    first_serve_points_match_lost = data.loc[data['loser_name'] == player_name, 'l_1stIn'].sum()
    second_serves_won_in_match_lost = data.loc[data['loser_name'] == player_name, 'l_2ndWon'].sum()
    total_service_points_match_lost = data.loc[data['loser_name'] == player_name, 'l_svpt'].sum()

    second_serves_won = second_serves_won_in_match_lost + second_serves_won_in_match_won

    total_second_serves = (total_service_points_match_lost + total_service_points_match_won) - (first_serve_points_match_lost + first_serve_points_match_won)

    second_serves_win_percentage = (second_serves_won / total_second_serves) * 100

    return second_serves_win_percentage

print(get_second_serve_win_percentage(data, 'Jannik Sinner'))
```

```python
def get_win_percentage(data, player_name):
    total_matches_won = len(data[(data['winner_name'] == player_name)])
    total_matches_lost = len(data[(data['loser_name'] == player_name)])

    return ((total_matches_won / (total_matches_lost + total_matches_won)) * 100)

print(get_win_percentage(data, 'Jannik Sinner'))
```

```python
def get_last_N_win_percentage(data, player_name, N):
    total_matches = data[(data['winner_name'] == player_name) | (data['loser_name'] == player_name)]
    last_N_matches = total_matches[-N:]

    wins_in_last_N_matches = len(last_N_matches[last_N_matches['winner_name'] == player_name])

    return (wins_in_last_N_matches / len(last_N_matches)) * 100
    

print(get_last_N_win_percentage(data, 'Jannik Sinner', 10))
```
